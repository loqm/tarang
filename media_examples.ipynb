### Dispersion Curves Fitted from Data Compared to the Literature.

Here we show how empirical data is used to fit dispersion curves. 

```python
# ---- LPCVD SiN ---- #
"""
Cite: Dispersion Optimization of Silicon Nitride Waveguides for
Efficient Four-Wave Mixing by Hong et. al., 2021.
https://doi.org/10.3390/photonics8050161
"""
from media import get_medium_dispersion

get_medium_dispersion(
    medium_key = "si3n4_hong",  # Get data from Hong et. al. (2021), LPCVD SiN
    output_path = "example_results/si3n4_hong_medium/dispersion.png",
    show = True,
    loglog = False,
    max_num_poles=8,
    force_zero_k = True
)
```

How good is the fit? TARANG's pole-residue fit (line) against the tabulated Hong *et al.* data (points), with the difference below. These points appear to be read from the paper's figure, so the fit cannot follow them more closely than about 1e-3.

# Comparing Dispersion Curves

```python
# ---- Get Dispersion Data ---- #
from media import get_medium_dispersion

# Si3N4, Luke et. al. 2015
Si3N4_Luke = get_medium_dispersion(
    medium_key = "si3n4",
    output_path = "example_results/Si3N4_Luke_2015/",
)

# Si3N4, Hong et. al., 2021
Si3N4_Hong = get_medium_dispersion(
    medium_key = "si3n4_hong",
    output_path = "example_results/Si3N4_Hong_2021/"
)

# Plotting
import matplotlib.pyplot as plt

plt.figure()
plt.plot(
    Si3N4_Luke["wl"], Si3N4_Luke["n"],
    color = "blue",
    label = "Luke et. al. 2015"
)

plt.plot(
    Si3N4_Hong["wavelength"], Si3N4_Hong["n"],
    color = "red",
    label = "Hong et. al. 2021"
)

plt.legend()
plt.show()
```

## Choosing a dataset

Fit each material over the band you will simulate, and stay inside the tabulated data: outside it the fitted index is an extrapolation, and TARANG warns. For 1550 nm work use `si3n4`, `sio2` and, for silicon, `si_franta` (the `si` dataset ends at 1.305 µm). References, ranges and recommendations for every dataset are in [media/data/SOURCES.md](media/data/SOURCES.md).
